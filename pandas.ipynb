```python
import pandas as pd

students = {
    'Name': ['Rahul', 'Priya', 'Amit', 'Neha'],
    'Semester': [3, 3, 5, 3],
    'CGPA': [8.5, 9.1, 7.8, 9.4]
    }
df = pd.DataFrame(students)
print("____CGPA coloum___")
print(df['CGPA'])

print("\n  students with cgpa greater then 8.0")
high_gpa = df[df['CGPA'] > 8.0]
print(high_gpa)
```

```python
import pandas as pd

data = {
    'Name': ['Rahul', 'Priya', 'Amit', 'Neha', 'Kiran', 'Sneha'],
    'Department': ['AI/ML', 'CSE', 'AI/ML', 'ECE', 'CSE', 'AI/ML'],
    'Semester': [3, 3, 5, 3, 5, 3],
    'CGPA': [8.5, 9.1, 7.8, 9.4, 8.2, 8.9],
}
df = pd.DataFrame(data)


dept_avg_cgpa = df.groupby('Department')['CGPA'].mean()

print(dept_avg_cgpa)
```
